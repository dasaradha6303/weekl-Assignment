Q34. For each city, compute three metrics in a single groupby using .agg() with a dictionary:
(a) Number of startups
(b) Total funding raised (sum of amount)
(c) Average funding per startup
Print the top 10 cities by number of startups.

```python
city_metrics=df.groupby('city').agg({'startup_name_':'count','amount':['sum','mean']})
city_metrics.columns=['num_startups','total_funding','avg_funding']
top_10_cities=city_metrics.sort_values(by='num_startups',ascending=False)
print(top_10_cities.head(10))
```

Q35. Interview-style. Which industry has the highest average funding, but only considering industries
with at least 15 startups? (Sparse industries bias the mean.) Print industry name, count, and average
funding.


```python
industry_metrics=df.groupby('industry').agg(
     startup_count=('startup_name_', 'count'),
        avg_funding=('amount', 'mean'))
filteed_indsteries=industry_metrics[industry_metrics['startup_count']>=15]
if not filteed_indsteries.empty:
    top_industry=filteed_indsteries.loc[filteed_indsteries['avg_funding'].idxmax()]
    print("industry",top_industry.name)
    print("count",top_industry['startup_count'])
    print("avg_funding",top_industry['avg_funding'])
```

Q39. Investor hub analysis. A foreign investor wants to know which Indian city has the highest TOTAL
funding raised, considering only cities with at least 20 startups (so the number is meaningful). Print city
name, count, and total funding.

```python
city_metrices=(df.dropna(subset=['city' ,'amount']).groupby('city').agg(
    startup_count=('startup_name_','count'),
    total_funding=('amount','sum')).reset_index())
filtered_cities=city_metrices[city_metrices['startup_count']>=20]
if not filtered_cities.empty:
    top_city=filtered_cities.loc[filtered_cities['total_funding'].idxmax()]
    print("city",top_city['city'])
    print("count",top_city['startup_count'])
    print("total_funding",top_city['total_funding'])
```

Q40. Market saturation. For each combination of (city, industry), count the number of startups. Print the
top 10 most-saturated combinations. This tells a new founder where competition is most intense – and
by elimination, where the gaps might be

```python
df_clea=df.dropna(subset=['city','industry'])
city_industry_counts=(
    df_clea.groupby(['city','industry'])
    .agg(startup_count=('startup_name_','count'))
    .reset_index())
top_10_saturated=city_industry_counts.sort_values(by='startup_count',ascending=False).head(10)
print(top_10_saturated)
```
